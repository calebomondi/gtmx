```python
import sys
from pathlib import Path

# Walk up from cwd to find project root (has pyproject.toml)
root = Path.cwd()
while root != root.parent:
    if (root / "pyproject.toml").exists():
        break
    root = root.parent

sys.path.insert(0, str(root))
```

```python
from langchain_qwq import ChatQwen
from langchain.agents import create_agent
from config import Config

from agents.composio.main import tools

cfg = Config()

llm = ChatQwen(
    model="qwen-plus-2025-07-28",
    api_key=cfg.dashscope_api_key,
    base_url=cfg.dashscope_base_url,
    timeout=300
)

agent = create_agent(
    tools=tools,
    model=llm
)

result = agent.invoke({"messages": [("user", "Send an email to omondimokua2@gmail.com with the subject 'Hello' and body 'Hello from Composio!'")]})
print(result["messages"][-1].content)
```
